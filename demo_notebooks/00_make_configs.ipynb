# Creating spinqick configuration files

Before running experiments with spinqick, you'll need a few configuration files.  Here we will demonstrate how to generate and edit them

```python
from spinqick.make_config import make_default_configs
```

## File Settings

The first step is to set a few environment variables which will define the desired file locations of spinqick configuration files, and the directory where you'd like to autosave data.  Spinqick will load these variables when you import `file_settings`. You'll need to set the `data_directory` variable to a file location on your machine where you can autosave data. The `hardware_config` and `dot_experiment_config` variables specify the location of the respective hardware and experiment json files created in the following steps. You can set these variables in a command line,  in Windows it would look like this:

`setx SPINQICK_DATA_DIRECTORY "your directory path"`

`setx SPINQICK_HARDWARE_CONFIG "your hardware config path"`

`setx SPINQICK_DOT_EXPERIMENT_CONFIG "your experiment config path"`

`setx SPINQICK_FILTER_CONFIG "your filter config path"`

After you set these variables, restart the window that your Jupyter notebook is running in.

## Generating Generic Config Files

The `make_default_configs` function will look for these env vars and generate default configs if desired. Set the arguments to be true to generate default files for hardware_config, dot_experiment_config and filter_config.

```python
make_default_configs(
    make_filter_cfg=True, make_experiment_cfg=True, make_hw_cfg=True, overwrite=False
)
```
